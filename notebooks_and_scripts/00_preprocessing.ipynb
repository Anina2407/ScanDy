```python
# change to the root directory of the project
import pandas as pd
import seaborn as sns
```

```python
import gzip
import shutil

# for each file seperately
path_gz = "data..."
path_csv = "data/..."
path_csv_clean = "data/.."

with gzip.open(path_gz, "rb") as f_in:
    with open(path_csv, "wb") as f_out:
        shutil.copyfileobj(f_in, f_out)
```

```python
df = pd.read_csv(path_csv)
bad_trials = (
    df.groupby("trial_id")["obj"]
      .apply(lambda s: s.isna().any())
)

# Keep only trials WITHOUT NaNs in obj
df_clean = df.loc[~df["trial_id"].isin(bad_trials[bad_trials].index)]

df_clean.to_csv(
    path_csv_clean,
    index=False
)
```
